```python
show_tables_query = "SHOW TABLES"
cursor.execute(show_tables_query)
results = cursor.fetchall()
for table in results:
    print(table)
```

```python
join_query = """
SELECT 
    Customers.FullName,
    Customers.ContactNumber,
    Orders.TotalCost
FROM Customers
INNER JOIN Orders
ON Customers.CustomerID = Orders.CustomerID
WHERE Orders.TotalCost > 20
"""

cursor.execute(join_query)

results = cursor.fetchall()

for row in results:
    print("Name:", row[0])
    print("Contact:", row[1])
    print("Total Cost:", row[2])
    print("------")

cursor.close()
connection.close()
```
